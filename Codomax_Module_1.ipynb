4.IF-ELSE-
This program checks whether the given input marks is greater than 35 or lesser than 35,where if the condition is true then print as pass and if the condition is false then it will print as fail.

```python
marks = int(input("Enter marks: "))

if marks >= 35:
    print("Pass")
else:
    print("Fail")
```

5.LOOPS-

```python
for i in range(1, 6):
    print(i)
```

**PYTHON PRACTICAL CONCEPTS-**

6.LISTS-
Here the list is used to store multiple values in one variable. This program finds the highest marks, lowest marks and average marks using max, min, sum and len functions.

```python
scores = [88, 92, 79, 95, 84]

highest = max(scores)
lowest = min(scores)
average = sum(scores) / len(scores)

print(f"Highest: {highest}, Lowest: {lowest}, Average: {average}")
```

7.DICTIONARY-
The dictionary stores information in key-value pairs.this program contains the keys as name,branch and marks and then the values will be entered which will be stored.

```python
student = {
    "name": "MONU",
    "branch": "AI & DS",
    "marks": 85
}

print(student["name"])
print(student["marks"])
```

8.Functions-
A function is a block of code used to perform a particular task, the add function takes two numbers, adds them and returns the result.

```python
def add(a, b):
    return a + b

print(add(10, 20))
```

9.FILE HANDLING-
File handling is used to store and read information from a file.the program creates a text file, writes a message into it and reads the message.

```python
with open("student.txt", "w") as f:
    f.write("Welcome to Codomax Internship")

with open("student.txt", "r") as f:
    print(f.read())
```

 DATA SCIENCE FUNDAMENTALS

**DATA SCIENCE-**

Data Science is the process of collecting, organizing, analyzing and understanding data to find useful information and make better decisions.

**2. Basic Concepts of Data Science**

1. Data Collection: Gathering information from different sources.
2. Data Cleaning: Removing incorrect or missing data.
3. Data Analysis: Studying data to find useful information.
4. Data Visualization: Representing data using graphs and charts.
5. Data Interpretation:Understanding the results and making decisions.

**3.APPLICATIONS-**

1. Healthcare: Predicting diseases and analyzing patient data.
2. Banking: Detecting fraud transactions.
3. Education: Analyzing student performance.
4. E-commerce: Recommending products to customers.
5. Transportation: Predicting traffic and improving routes.
